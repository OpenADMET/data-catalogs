```python
from openadmet.toolkit.database import rcsb_utils as rut
```

This file downloads and saves raw files from [RCSB PDB](https://www.rcsb.org/) using functions available in [openadmet-toolkit](https://github.com/OpenADMET/openadmet-toolkit).

# Query PDB IDs

To start define your target of interest. This target should match a key in the internal dictionary UNIPROT_IDs. If you already have the Uniprot ID for your target, this can act as input into the query function below, get_pdb_ids. 

```python
rut.UNIPROT_IDS.keys()
```

```python
target = "PXR"
```

```python
pdb_ids = rut.get_pdb_ids(rut.UNIPROT_IDS[target])
```

The available functionality allows for downloading mmCIF or PDB file types. The code is defaulted for the cif format, as pdb is depricated. If you are interested in PDB files, however, conversion code from mmCIF to PDB is available for newer files that do not have a PDB file associated.  
An example for both downloading mmCIF and PDB are shown below. 

```python
# Download mmCIF

output = f'example_output/cif_download_test/'
for _id in pdb_ids:
    pdb_response = rut.get_rcsb_url(_id)
    pdb_text = rut.download_rcsb_file(_id, outdir=output)
```

```python
# Download PDB

output = f'example_output/pdb_download_test/'
for _id in pdb_ids:
    pdb_response = rut.get_rcsb_url(_id)
    try:
        pdb_text = rut.download_rcsb_file(_id, fmt='pdb', outdir=output)
    except ValueError:
        print(f"PDB not available for {_id}. Trying to download mmCIF and convert to PDB.")
        cif_text = rut.download_rcsb_file(_id, fmt='cif', outdir=output, get_text=True)
        pdb_text = rut.convert_cif_to_pdb_gemmi(cif_text)
        print(f"Saving {_id} PDB from mmCIF...")
        # Conversion function does not save file, so you must specify your direct path below
        with open(f"{output}/{_id}.pdb", "w") as f:
            f.write(pdb_text)
```
