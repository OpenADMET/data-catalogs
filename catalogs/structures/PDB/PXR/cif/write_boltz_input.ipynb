```python
import pandas as pd
from pathlib import Path
```

Read the data generated by `pxr_preprocess.ipynb`

```python
pxr_df_clean = pd.read_parquet("pxr_clean_ligand_info_and_sequences.parquet")
```

Define a simple a simple template for Boltz-2 input 

```python
buff = """version: 1  # Optional, defaults to 1
sequences:
  - protein:
      id: A
      sequence: PROTEIN_SEQUENCE
  - ligand:
      id: B
      smiles: 'LIGAND_SMILES'
properties:
  - affinity:
      binder: B"""
```

Loop over the structures and write the Boltz-2 input files

```python
Path("Boltz_inputs").mkdir(exist_ok=True)
for pdb,smiles,seq in pxr_df_clean[["pdb","SMILES","sequence"]].values:
    new_buff = buff.replace("PROTEIN_SEQUENCE",seq).replace("LIGAND_SMILES",smiles)
    with open(f"Boltz_inputs/{pdb}.yaml","w") as ofs:
        ofs.write(new_buff)       
```
