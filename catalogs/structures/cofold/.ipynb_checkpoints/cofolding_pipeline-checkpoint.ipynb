## Protein Co-Folding Pipeline
This notebook begins with a CSV file containing the names and UniProt IDs of Avoid-ome proteins.  The script then retrieves the UniProt IDs for the corresponding proteins and writes input files for Boltz-2 and OpenFold-3. 

```python
import pandas as pd
from get_protein_sequence import get_protein_sequence
from pathlib import Path
```

```python
df = pd.read_csv("uniprot_ids.csv")
```

```python
df
```

```python
df['sequence'] = df["UniProt ID"].apply(get_protein_sequence)
```

```python
df.to_csv("uniprot_ids_with_sequence.csv",index=False)
```

```python
def generate_boltz_input(sequence, output_filename):
    buff = """version: 1  # Optional, defaults to 1
sequences:
  - protein:
      id: A
      sequence: PROTEIN_SEQUENCE
"""
    with open(output_filename,"w") as ofs:
        print(buff.replace("PROTEIN_SEQUENCE",seq),file=ofs)

def generate_of3_input(sequence, output_filename):
    buff = """{
    	"queries":{
    		"NAME": {
    			"chains": [
    				{
    					"molecule_type": "protein",
    					"chain_ids": ["A"],
    					"sequence": "PROTEIN_SEQUENCE"
    				}
    			]
    		}
    	}
    }"""
    with open(output_filename,"w") as ofs:
        print(buff.replace("PROTEIN_SEQUENCE",seq),file=ofs)
```

```python
dir_name = "Boltz_inputs"
path = Path(f"./{dir_name}")
path.mkdir(parents=True, exist_ok=True)
for name, uniprot, seq in df.values:
    generate_boltz_input(seq,f"{dir_name}/{name}.yaml")
```

```python
dir_name = "OF3_inputs"
path = Path(f"./{dir_name}")
path.mkdir(parents=True, exist_ok=True)
for name, uniprot, seq in df.values:
    generate_of3_input(seq,f"{dir_name}/{name}.json")
```
