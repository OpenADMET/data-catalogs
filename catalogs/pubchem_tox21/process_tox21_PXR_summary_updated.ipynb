```python
from openadmet.toolkit.chemoinformatics.data_curation import DataProcessing
```

```python
file = "./processed/PXR/processed_AID_1347033_datatable.parquet"

df = DataProcessing.read_file(file)
df
```

```python
processor = DataProcessing()
# data = processor.standardize_smiles_and_convert(data=df)
# data
```

```python
df.columns
```

```python
data = processor.get_pac50(data=df, pac50_col='Agonist Potency (uM)', input_unit="uM", activity_type="EC50")
data
```

```python
nan_df = data.loc[data['OPENADMET_LOGAC50'].isna(), :]
nan_df["PUBCHEM_ACTIVITY_SCORE"].value_counts()
```

```python
import matplotlib.pyplot as plt

plt.hist(x=nan_df["PUBCHEM_ACTIVITY_SCORE"])
plt.xlabel("PUBCHEM_ACTIVITY_SCORE")
plt.title("Distribution of compounds with OPENADMET_LOGAC50=NaN")
plt.show()
```

```python
plt.hist(x=data["PUBCHEM_ACTIVITY_SCORE"])
plt.xlabel("PUBCHEM_ACTIVITY_SCORE")
plt.title("Distribution of all compounds PXR Tox21")
plt.show()
```
